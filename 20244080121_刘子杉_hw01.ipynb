```python
# 1.say_hi：无参，打印 Hello, world!
def say_hi():
    print("Hello, world!")

# 2.goat_pad：接收字符串，前后拼接goat后打印
def goat_pad(s):
    print("goat" + s + "goat")

#3.print_n：字符串s，整数n；打印s共n次，每行1个
def print_n(s, n):
    for _ in range(n):
        print(s)

# --------测试调用--------
print("-----测试 say_hi-----")
say_hi()

print("\n-----测试 goat_pad('apple')-----")
goat_pad("apple")

print("\n-----测试 print_n('test',3)-----")
print_n("test", 3)
```

```python
#1.实现gcd欧几里得算法
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

#2.测试三组数字对
print("gcd(2018,2019) =", gcd(2018, 2019))
print("gcd(1200,300) =", gcd(1200, 300))
print("gcd(5040,60) =", gcd(5040, 60))

#3.测试负数参数
print("\n负数参数测试 gcd(-24, -18):", gcd(-24, -18))
print("负数参数测试 gcd(-24, 18):", gcd(-24, 18))
```

```python
import math

#1.euler_limit：极限公式 (1+1/n)^n
def euler_limit(n):
    return (1 + 1 / n) ** n

#2.euler_infinite_sum：前n项求和 e = sum 1/k!
def euler_infinite_sum(n):
    total = 0.0
    fact = 1  # 0! = 1
    for k in range(n):
        if k == 0:
            fact = 1
        else:
            fact = fact * k
        total = total + 1.0 / fact
    return total

#3.euler_approx：误差ϵ，循环累加直到和与真实e差距小于ϵ
def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - true_e) >= epsilon:
        if k == 0:
            fact = 1
        else:
            fact = fact * k
        total += 1.0 / fact
        k += 1
    return total

#4.print_euler_sum_table：k从1~n，打印每一项euler_infinite_sum(k)
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")

# --------测试--------
print("真实e值 math.exp(1) =", math.exp(1))
print("\neuler_limit测试：")
print(f"euler_limit(1000) = {euler_limit(1000)}")

print("\neuler_infinite_sum测试：")
print(f"euler_infinite_sum(0) = {euler_infinite_sum(0)}")
print(f"euler_infinite_sum(1) = {euler_infinite_sum(1)}")
print(f"euler_infinite_sum(4) = {euler_infinite_sum(4)}")

print("\neuler_approx(0.001) =", euler_approx(0.001))

print("\n-----print_euler_sum_table(6)-----")
print_euler_sum_table(6)
```

```python
#1.is_power_of_2 判断n是否为2的幂，只用除法、取模，不能sqrt
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

#2.is_power(b,n) 判断 n 是否是 b 的幂
def is_power(b, n):
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

# --------测试用例--------
print("-----is_power_of_2测试-----")
test_2 = [1,2,4,8,9,10,16]
for num in test_2:
    print(f"{num} 是2的幂? {is_power_of_2(num)}")

print("\n-----is_power(b,n)测试-----")
test_cases = [(3,9), (2,16), (5,125), (2,7), (4, 64)]
for b,n in test_cases:
    print(f"is_power({b},{n}) = {is_power(b,n)}")
```
