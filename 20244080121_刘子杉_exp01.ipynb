```python
#1.score_level函数
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须在0~100")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

print("-----score_level测试-----")
test_scores = [95,88,72,55,-5,101]
for sc in test_scores:
    try:
        res = score_level(sc)
        print(f"{sc}分 → {res}")
    except ValueError as e:
        print(f"{sc}分 → 异常:{e}")


#2.for、while求1~100和
print("\n-----1~100求和-----")
sum_for = 0
for i in range(1,101):
    sum_for +=i
print(f"for循环求和:{sum_for}")

sum_while = 0
i=1
while i<=100:
    sum_while +=i
    i +=1
print(f"while循环求和:{sum_while}")


#3.迭代阶乘、递归阶乘
def fact_iter(n):
    if n <0:
        raise ValueError("非负整数")
    res =1
    for i in range(2,n+1):
        res *=i
    return res

def fact_rec(n):
    if n <0:
        raise ValueError("非负整数")
    if n ==0 or n ==1:
        return 1
    return n * fact_rec(n-1)

print("\n-----阶乘测试-----")
for n in [0,1,5,10]:
    print(f"n={n},迭代阶乘={fact_iter(n)},递归阶乘={fact_rec(n)}")


#4.summary函数 返回(最小,最大,平均)，元组解包
def summary(values):
    if len(values)==0:
        raise ValueError("列表不能为空")
    minv = min(values)
    maxv = max(values)
    avgv = sum(values)/len(values)
    return (minv, maxv, avgv)

print("\n-----summary函数-----")
num_list = [12,45,33,7]
min_num, max_num, avg_num = summary(num_list) #元组解包
print(f"min={min_num},max={max_num},avg={avg_num:.2f}")


#5.sum_all(*args)可变参数
def sum_all(*args):
    return sum(args)

print("\n-----sum_all可变参数-----")
print(f"无参调用sum_all(): {sum_all()}")
print(f"单参数sum_all(10): {sum_all(10)}")
print(f"多参数sum_all(1,2,3,4,5): {sum_all(1,2,3,4,5)}")
```

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

#1.zip构建字典，items遍历
stu_dict = dict(zip(names, scores))
print("1.zip构建字典 items遍历：")
for name,sc in stu_dict.items():
    print(f"姓名:{name},成绩:{sc}")

#2.enumerate序号从1开始
print("\n2.enumerate输出序号、姓名、成绩")
for idx,(name,sc) in enumerate(zip(names,scores), start=1):
    print(f"序号{idx}，姓名{name}，成绩{sc}")

#3.列表推导式
print("\n3.列表推导式")
pass_name = [name for name,sc in zip(names,scores) if sc>=60]
score_square = [s**2 for s in scores]
print(f"及格学生姓名：{pass_name}")
print(f"成绩平方列表：{score_square}")

#4.统计
count = len(scores)
max_sc = max(scores)
min_sc = min(scores)
avg_sc = sum(scores)/count
pass_count = len([s for s in scores if s>=60])
pass_rate = pass_count / count
print("\n4.统计结果")
print(f"总人数:{count},最高分:{max_sc},最低分:{min_sc}")
print(f"平均分:{avg_sc:.2f},及格率:{pass_rate:.1%}")

#5.别名赋值 vs copy()，独立演示列表，不修改scores
print("\n5.列表别名与copy演示")
demo_list = [10,20,30,40]
alias = demo_list
copied = demo_list.copy()
alias[1] = 999
print(f"原demo_list:{demo_list}")
print(f"alias(别名):{alias}")
print(f"copied副本:{copied}")

#6.元组不可修改，不等长zip
print("\n6.元组与不等长zip")
stu_tuple = ("Tom",77)
try:
    stu_tuple[1]=88
except TypeError as e:
    print(f"元组赋值报错:{e}")

listA = [1,2,3]
listB = [10,20]
z = list(zip(listA,listB))
print(f"不等长序列zip结果:{z}")
```

```python
import os
import pickle

#1.查看工作目录，创建文件夹
cwd = os.getcwd()
print(f"当前工作目录：{cwd}")
data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print(f"数据目录路径:{data_dir},目录是否存在:{os.path.exists(data_dir)}")

#2.写入students.txt，再读取
txt_path = os.path.join(data_dir,"students.txt")
content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
#写文件
with open(txt_path,mode="w",encoding="utf-8") as f:
    f.write(content)
#读文件
with open(txt_path,mode="r",encoding="utf-8") as f:
    read_text = f.read()
print("\n读取students.txt全部内容：")
print(read_text)

#3.parse_student函数
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts)!=2:
        raise ValueError(f"字段数量错误，分割得到{len(parts)}个字段")
    name_str, score_str = parts
    name_str = name_str.strip()
    score_str = score_str.strip()
    if len(name_str)==0:
        raise ValueError("姓名为空")
    try:
        score_int = int(score_str)
    except ValueError:
        raise ValueError(f"成绩无法转为整数:{score_str}")
    if not (0 <= score_int <=100):
        raise ValueError(f"成绩超出0‑100范围:{score_int}")
    return {"name":name_str, "score":score_int}

#4.逐行处理，捕获异常
valid_records = []
error_log = []
with open(txt_path,"r",encoding="utf-8") as f:
    lines = f.readlines()
for line_num, one_line in enumerate(lines, start=1):
    if line_num ==1:
        continue #跳过表头
    try:
        rec = parse_student(one_line)
        valid_records.append(rec)
    except ValueError as err:
        error_log.append({"line":line_num,"err":str(err)})

print("\n---有效记录---")
print(valid_records)
print("\n---错误日志(行号+原因)---")
for e in error_log:
    print(e)

#5.统计写入report.txt w模式 和 a追加模式
valid_num = len(valid_records)
invalid_num = len(error_log)
if valid_num>0:
    avg_score = sum(r["score"] for r in valid_records)/valid_num
    pass_cnt = sum(1 for r in valid_records if r["score"]>=60)
    pass_ratio = pass_cnt / valid_num
else:
    avg_score=0
    pass_ratio=0

report_path = os.path.join(data_dir,"report.txt")
#w模式覆盖写
with open(report_path,"w",encoding="utf-8") as fw:
    fw.write(f"有效记录数:{valid_num}\n")
    fw.write(f"无效记录数:{invalid_num}\n")
    fw.write(f"平均分:{avg_score:.2f}\n")
    fw.write(f"及格率:{pass_ratio:.1%}\n")

#a模式追加
with open(report_path,"a",encoding="utf-8") as fa:
    fa.write("=====这是a模式追加新增文字=====\n")

#读取report查看结果
with open(report_path,"r",encoding="utf-8") as fr:
    report_content = fr.read()
print("\nreport.txt(w覆盖+a追加后内容):")
print(report_content)

#6.pickle保存读取
pkl_path = os.path.join(data_dir,"students.pkl")
#dump保存
with open(pkl_path,"wb") as fpw:
    pickle.dump(valid_records, fpw)
#load读取
with open(pkl_path,"rb") as fpr:
    load_data = pickle.load(fpr)

print("\npickle加载后数据与原数据是否相等：", load_data == valid_records)

#8.捕获FileNotFoundError
print("\n---捕获不存在文件异常测试---")
not_exist_file = os.path.join(data_dir,"no_file.txt")
try:
    with open(not_exist_file,"r",encoding="utf-8") as f:
        pass
except FileNotFoundError as e:
    print(f"捕获FileNotFoundError:{e}")
```

```python
import copy

#1.定义Student类
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score
    def is_passed(self):
        return self.score >=60
    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"

#2.把实验4的valid_records生成Student对象
print("-----Student类实例-----")
for rec in valid_records:
    s_obj = Student(rec["name"], rec["score"])
    print(s_obj, f"是否及格:{s_obj.is_passed()}")

#3.子类GraduateStudent继承
class GraduateStudent(Student):
    def __init__(self,name,score,research_topic):
        super().__init__(name,score)
        self.research_topic = research_topic
    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},研究方向={self.research_topic})"

g_stu = GraduateStudent("Zhang",91,"大模型")
print("\n-----子类GraduateStudent-----")
print(g_stu)
print(f"是否及格:{g_stu.is_passed()}")

#4.直接赋值 copy.copy copy.deepcopy对比嵌套对象
print("\n-----嵌套对象复制对比-----")

#①直接赋值
ori1 = {"name": "Alice", "scores": [85, 90]}
obj1 = ori1
obj1["scores"][0]= 999
print("【直接赋值】")
print(f"原始:{ori1}")
print(f"赋值对象:{obj1}")

#重建原始对象
ori2 = {"name": "Alice", "scores": [85, 90]}
obj2 = copy.copy(ori2)
obj2["scores"][0]=999
print("\n【copy.copy浅拷贝】")
print(f"原始:{ori2}")
print(f"copy对象:{obj2}")

#重建原始对象
ori3 = {"name": "Alice", "scores": [85, 90]}
obj3 = copy.deepcopy(ori3)
obj3["scores"][0]=999
print("\n【copy.deepcopy深拷贝】")
print(f"原始:{ori3}")
print(f"deepcopy对象:{obj3}")
```
